```python
# 9 Accept 20 values from user and save it in list.
values = []
print("Enter 20 values:")
for i in range(20):
    val = int(input(f"Value {i+1}: "))
    values.append(val)

    
# a) count similar elements of list and print their index value 
seen = {}
for idx, val in enumerate(values):
    seen.setdefault(val, []).append(idx)

for val, indices in seen.items():
    if len(indices) > 1:
        print(f"Element {val} appears {len(indices)} times at indices: {indices}")

        
# b) count even and odd values of list  
evens = sum(1 for x in values if x % 2 == 0)
odds = len(values) - evens
print(f"\n--- b) Even & Odd Counts ---")
print(f"Even values: {evens}, Odd values: {odds}")


# c) count positive and negative values of list. 
pos = sum(1 for x in values if x > 0)
neg = sum(1 for x in values if x < 0)
print(f"\n--- c) Positive & Negative Counts ---")
print(f"Positive values: {pos}, Negative values: {neg}")
```

```python
# 10 Accept 10 values from user and save it in list. Perform following operations on it   
# a) Sort list in ascending order using sorted() function and display sorted list 
# b) sort list in descending order using sort() function 
# c) display length of list 
values=[]
for i in range(10):
    val=int(input(f"value {i+1}: "))
    values.append(val)
ascnd= sorted(values)  
dscnd=sorted(values, reverse=True)
lngt=len(values)
print("Original list:", values)
print("Ascending list:", ascnd)
print("Dscending list:", dscnd)
print("Lenght of list:", lngt)
```

```python
#12. Generate an acronym from a phrase (all uppercase)
phrase = input("Enter a phrase: ")
words = phrase.split()
letters = []
for i in words:
    letters.append(i[0].upper())
arc = ".".join(letters)  
print(f"the acronym form of the phrase({phrase}) is ({arc})")  
```
